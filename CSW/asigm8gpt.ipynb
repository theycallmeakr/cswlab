```

# Q1
import numpy as np

v1 = np.random.randint(1, 50, size=20)
v2 = np.random.randint(1, 50, size=20)

def analyze_vector(v):
    return np.max(v), np.min(v), np.sum((v > np.min(v)) & (v < np.max(v)))

print(v1, analyze_vector(v1))
print(v2, analyze_vector(v2))
```

```

# Q2
v1 = np.random.randint(1, 10, size=(3,3))
v2 = np.random.randint(1, 10, size=(3,3))

v1f = v1.flatten()
v2f = v2.flatten()

v1n = v1f / np.linalg.norm(v1f)
v2n = v2f / np.linalg.norm(v2f)

angle_rad = np.arccos(np.dot(v1n, v2n))
angle_deg = np.degrees(angle_rad)

print(angle_rad, angle_deg)
```

```

# Q5
sugar = np.array([85,110,140,95,130,160,100])
result = np.empty(sugar.shape, dtype=object)

result[sugar < 100] = "Normal"
result[(sugar >= 100) & (sugar <= 139)] = "Pre-Diabetic"
result[sugar >= 140] = "Diabetic"

print(result)
```

```

# Q6
data = np.random.randn(1000)
mean = data.mean()
std = data.std()

dist = {
    "1_std": np.sum((data >= mean-std) & (data <= mean+std)),
    "2_std": np.sum((data >= mean-2*std) & (data <= mean+2*std)),
    "3_std": np.sum((data >= mean-3*std) & (data <= mean+3*std))
}
print(dist)
```

```

# Q7 (interactive)
import numpy as np
n = int(input("Enter size: "))
hidden = np.random.randint(1,10,size=n)
trials = []

for i in range(n):
    c = 0
    while True:
        g = int(input(f"Guess row {i}: "))
        c += 1
        if g == hidden[i]:
            trials.append(c)
            break
print(trials)
```

```

# Q8
mat = np.zeros((4,5),dtype=int)
for i in range(4):
    mat[i] = np.random.randint(i+1, i+9, 5)

result = {
    "mean": "row-wise" if mat.mean(axis=1).mean() > mat.mean(axis=0).mean() else "column-wise",
    "std": "row-wise" if mat.std(axis=1).mean() > mat.std(axis=0).mean() else "column-wise"
}
print(result)
```
