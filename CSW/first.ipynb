```python
import struct
number = 1025
big_endian = struct.pack('>H', number)
little_endian = struct.pack('<H', number)
print(f"orignal number: {number}")
print("decimal representation in big-endian:", list(big_endian))
print(f"\n Big-endian ('>H') raw bytes: {big_endian}")
print(f"   Byte 1(MSb): {hex( big_endian[0])}, Byte 2(LSb): {hex(big_endian[1])}")
print(f"\n little -endian ('<H') raw bytes: {little_endian}")
print(f"   Byte 1(LSb): {hex( little_endian[0])}, Byte 2(MSb): {hex(little_endian[1])}")
```
