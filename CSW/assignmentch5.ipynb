```python
#assignment chapter 6 question 1
def filter_emails(lst):
    valid = []
    for email in lst:
        # must contain exactly one '@'
        if email.count('@') != 1:
            continue
        username, domain_part = email.split('@')

        # username rules
        if not username:
            continue
        if not(username[0].isalnum()):    # must start with letter or digit
            continue
        for ch in username:
            if not (ch.isalnum() or ch in ['.', '_']):
                break
        else:
            # domain part must contain exactly one dot
            if domain_part.count('.') != 1:
                continue

            domain, tld = domain_part.split('.')

            # domain must be letters only
            if not domain.isalpha():
                continue

            # TLD must be 2–4 letters
            if not (2 <= len(tld) <= 4 and tld.isalpha()):
                continue

            valid.append(email)

    return valid


# Example usage:
emails = ["student123@gmail.com", "teacher.name@soa.edu", "abc.xyz@abc.in",
          "xyz#12@abc.com", "bad@1n.com"]

print(filter_emails(emails))
```

```python
#q2
import re

def verify_passwd(s):

    # Rule 1: At least 8 characters
    rule1 = re.match(r'.{8,}$', s)

    # Rule 2: Only allowed characters (letters, digits, @ # $ % ^ & * !)
    rule2 = re.match(r'^[A-Za-z0-9@#$%^&*!]+$', s)

    # Rule 3: At least one digit
    rule3 = re.search(r'[0-9]', s)

    # Rule 4: At least one letter
    rule4 = re.search(r'[A-Za-z]', s)

    # Rule 5: At least one special character
    rule5 = re.search(r'[@#$%^&*!]', s)

    return bool(rule1 and rule2 and rule3 and rule4 and rule5)
passwords = [
    "abc123!!",
    "pass1234",
    "PASSWORD@1",
    "short1!",
    "Valid@123",
    "NoSpecial123",
    "NoDigit!!!",
]

for p in passwords:
    print(p, "-", verify_passwd(p))
```

```python
#q3
import re

def normalize_phones(s):
    # pattern to match: +91-, (91), 0091, 91 followed by 10 digits
    pattern = r'(?:\+91-|\(91\)\s*|0091\s*|91\s*)(\d{10})'

    def repl(m):
        return "+91-" + m.group(1)

    return re.sub(pattern, repl, s)


text = "Contact: +91-9876543210, Office: (91) 98765 43210, Home: 0091 9876543210"
print(normalize_phones(text))
```

```python
#q5
import re

def match_cvc(word):
    return bool(re.fullmatch(r'c[aeiou][^aeiou]', word))


# Testing
tests = ["cat", "cit", "cot", "cut", "caa", "coo", "c9t"]
for t in tests:
    print(t, "→", match_cvc(t))
```

```python
#8
import re

def extract_date_parts(s):
    m = re.search(r'(\d{2})-(\d{2})-(\d{4})', s)
    if not m:
        return None
    return (m.group(0),
            m.group(1),
            m.group(2),
            m.group(3),
            m.span(),
            m.lastindex)


print(extract_date_parts("Backup 05-11-2025 complete"))
```

```python
#9
import re

def find_upper_digit(s):
    m = re.search(r'([A-Z]).*\d', s)
    if m:
        return (m.group(1), m.start(1))
    return None


print(find_upper_digit("a B blah 9"))
```

```python
#10
import re

def find_repeated_letters(s):
    pattern = r'(.)\1+'
    for m in re.finditer(pattern, s):
        print("Sequence:", m.group(0),
              "| Letter:", m.group(1),
              "| Span:", m.span())


# Example
find_repeated_letters("aabbccdd xyz bbb zzzz hello")
```
